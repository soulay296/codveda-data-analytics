Data Cleaning and Preprocessing

```python
#cleaning churn-bigml-20.csv dataset
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('churn-bigml-20.csv')
display(df)

print(df.duplicated().sum())
print(df.isnull().sum())
df['International plan']=df['International plan'].replace({"No":0,"Yes":1})
df['Voice mail plan']=df['Voice mail plan'].replace({"No":0,"Yes":1})
df["Churn"]=df["Churn"].astype(int)
df.to_csv("cleaned_churned_20.csv")
```

```python
#cleaning churn-bigml-80.csv dataset
df=pd.read_csv('churn-bigml-80.csv')
display(df)
print(df.duplicated().sum())
print(df.isnull().sum())
df['International plan']=df['International plan'].replace({"No":0,"Yes":1})
df['Voice mail plan']=df['Voice mail plan'].replace({"No":0,"Yes":1})
df["Churn"]=df["Churn"].astype(int)
df.to_csv("cleaned_churned_80.csv")
```
